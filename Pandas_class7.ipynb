## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python
import pandas as pd

# Dataset 1: Product details
df1 = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"],
    "Price": [55000, 800, 1500, 12000, 2500],
    "Stock": [10, 50, 30, 15, 20]
})

# Dataset 2: More products (same columns as df1 → good for vertical stacking)
df2 = pd.DataFrame({
    "Product": ["Headphones", "Speaker", "Tablet", "Charger", "USB Cable"],
    "Price": [3000, 4500, 25000, 1200, 500],
    "Stock": [25, 12, 8, 40, 100]
})

# Dataset 3: Additional info (different columns → good for horizontal stacking)
df3 = pd.DataFrame({
    "Brand": ["Dell", "Logitech", "HP", "Samsung", "Sony"],
    "Rating": [4.5, 4.2, 4.3, 4.6, 4.1],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Accessories"]
})
```

```python
pd.concat([df1, df2], ignore_index=True)
```

```python
pd.concat([df1, df3], axis=1)
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
products = pd.concat([df1,df2],ignore_index=True)
```

```python
products
```

```python
sales = pd.DataFrame({
    "Product": ["Laptop", "Keyboard", "Monitor", "Tablet", "Speaker"],
    "Units_Sold": [5, 15, 8, 3, 6],
    "Revenue": [275000, 22500, 96000, 75000, 27000]
})
```

```python
# For merging based on data available 
pd.merge(products, sales)
```

```python
# For merging based on data available 
pd.merge(products, sales, on="Product", how="inner")
```

```python
# For merging based on data available 
pd.merge(products, sales, on="Product", how="left")
```

```python
# For merging based on data available 
pd.merge(products, sales, on="Product", how="right")
```

```python
# For merging based on data available 
pd.merge(products, sales, on="Product", how="outer")
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
import pandas as pd

projects = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 105, 108],
    "Project": ["Alpha", "Beta", "Gamma", "Delta", "Omega"]
})

# DataFrame 4: Employee ID and performance rating
performance = pd.DataFrame({
    "Rating": [4.5, 3.8, 4.9, 4.2, 3.5]
}, index=[101, 102, 103, 105, 108])
```

```python
proj_df = projects.set_index("EmployeeID")
```

```python
proj_df.join(performance)
```

## Mini Practice
1. Create three DataFrames: Products, Sales, and Revenue.
2. Concatenate them vertically and horizontally.
3. Merge Sales and Revenue using different join types.
4. Join Products with Sales based on index or a key.
5. Analyze missing values after joins.

```python
import pandas as pd
# 1. Create three DataFrames: Products, Sales, and Revenue

# DataFrame 1: Products
products = pd.DataFrame({
    "Product_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam", "Headphones"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Accessories", "Audio"]
})

products.set_index("Product_ID",inplace=True)

# DataFrame 2: Sales
sales = pd.DataFrame({
    "Product_ID": [101, 102, 104, 107, 108],
    "Units_Sold": [5, 20, 8, 12, 15],
    "Sales_Region": ["Mumbai", "Delhi", "Pune", "Chennai", "Bangalore"]
})
sales.set_index("Product_ID", inplace=True)

# DataFrame 3: Revenue
revenue = pd.DataFrame({
    "Product_ID": [101, 103, 104, 105, 109],
    "Revenue": [275000, 45000, 96000, 50000, 75000],
    "Profit": [35000, 8000, 15000, 12000, 18000]
})
revenue.set_index("Product_ID", inplace=True)

# 2. Concatenate them vertically and horizontally
# Write your code here
full_df = pd.concat([products, sales, revenue], axis=1)
print(full_df)

# 3. Merge Sales and Revenue using different join types (inner, left, right, outer)
# Write your code here
print(pd.merge(sales, revenue, on="Product_ID", how="inner"))
print(pd.merge(sales, revenue, on="Product_ID", how="left"))
print(pd.merge(sales, revenue, on="Product_ID", how="right"))
print(pd.merge(sales, revenue, on="Product_ID", how="outer"))

# 4. Join Products with Sales based on index or a key
# Write your code here
print(products.join(sales, on="Product_ID"))

# 5. Analyze missing values after joins
# Use isna(), info(), or describe() to explore gaps
# Write your code here
print(full_df.isna().sum())
print(full_df.info())
print(full_df.describe())
```
