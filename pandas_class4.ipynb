## 1. Boolean Indexing & Conditional Selection
- Use comparison operators to build conditions (e.g., `df['col'] > 50`).
- Pass the condition inside `df[...]` to get matching rows.
- Combine multiple conditions with `&` (AND), `|` (OR), and `~` (NOT).
- Parentheses are required when combining conditions.

```python
import pandas as pd
import numpy as np
data = np.array([
    [101,  "Alice",   25,  "Delhi",     45000],
    [102,  "Bob",     30,  "Mumbai",    55000],
    [103,  "Charlie",  None, "Pune",    60000],
    [104,  "David",   40,  "Bangalore", None],
    [105,  "Alice",   25,  "Delhi",     45000],
    [106,  "Eva",     35,  "Mumbai",    70000],
    [107,  "Frank",   None, "Pune",     50000],
    [108,  "Grace",   28,  "Delhi",     65000],
    [109,  "Bob",     30,  "Mumbai",    55000],
    [110,  "Henry",   45,  None,        80000]
], dtype=object)
```

```python
df = pd.DataFrame(data, columns=["ID", "Name", "Age", "City", "Salary"])
```

```python
df
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Salary"] = df["Salary"].astype(int)
```

```python
df
```

```python
print(df["Age"] > 30) # Will give the column with boolean values
# To get the rows of data
df[df["Age"] > 30]
```

```python
# We can combine condition using logical operators &, | and ~
# Paranthesis is a must for combining logical operators
df[(df["Age"] > 30) & (df["City"] == "Mumbai")]
```

## 2. Using `query()` for Cleaner Filtering
- `query()` allows filtering rows with a string expression.
- It's concise and readable for complex conditions.
- Variables can be passed using the `@` symbol.

```python
# Its similar to conditional filtering but done in strings
df.query("Age > 30 and City == 'Bangalore'")
```

```python
# we can make use of variables here using @
salaryStart = 40000
salaryEnd = 60000
df.query("Salary > @salaryStart and Salary < @salaryEnd")
```

## 3. String Operations on Columns
- Use the `.str` accessor for vectorized string methods.
- Common methods: `contains`, `replace`, `startswith`, `endswith`, `lower`, `upper`.
- Useful for cleaning or filtering text data.

```python
df["Name"].str.lower()
```

```python
df["Name"].str.upper()
```

```python
df["Name"].str.startswith("A")
```

```python
df["Name"].str.endswith("e")
```

```python
df["Name"].str.contains("a")
```

```python
# It means ignore the case
df["Name"].str.contains("a", case = False)
```

```python
df["Name"].str.replace("Alice", "Angel")
```

## Mini Practice
1. Build a DataFrame with columns such as Name, Age, City, Score.
2. Use boolean indexing to select rows where Age > 25.
3. Filter rows where City is either 'Delhi' or 'Mumbai'.
4. Use query() to find rows with Score >= 85 and Age < 35.
5. Clean a text column using str.replace or str.lower.

```python
import pandas as pd

# 1. Build a DataFrame with columns: Name, Age, City, Score
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 22, 35, 28],
    "City": ["Mumbai", "Delhi", "Pune", "Bangalore", "Mumbai"],
    "Score": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

# 2. Use boolean indexing to select rows where Age > 25
# Write your code here
print(df[df["Age"] > 25])

# 3. Filter rows where City is either 'Delhi' or 'Mumbai'
# Write your code here
print(df[(df["City"] == "Mumbai") | (df["City"] == "Delhi")])

# 4. Use query() to find rows with Score >= 85 and Age < 35
# Write your code here
print(df.query("Age < 35 and Score >= 85"))

# 5. Clean a text column using str.replace or str.lower
# Example: convert City names to lowercase or replace "delhi" with "New Delhi"
# Write your code here
df["City"] = df["City"].str.lower()
print(df)
df["City"] = df["City"].str.replace("delhi", "new delhi")
print(df)
```
